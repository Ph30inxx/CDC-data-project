```python
import numpy as np
import pandas as pd
from pathlib import Path

from sklearn.metrics import mean_squared_error, r2_score
from xgboost import XGBRegressor

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from PIL import Image

import torchvision.models as models
from torchvision.models import ResNet18_Weights  # weights API is documented. [web:41]

proc = Path("data/processed")

X_train_tab = np.load(proc / "X_train_tab.npy")
X_val_tab   = np.load(proc / "X_val_tab.npy")
X_test_tab  = np.load(proc / "X_test_tab.npy")

y_train = np.load(proc / "y_train.npy")
y_val   = np.load(proc / "y_val.npy")

# Train on log scale
y_train_log = np.log1p(y_train)
y_val_log   = np.log1p(y_val)

train_ids = pd.read_csv(proc / "train_ids.csv")["id"].astype(str).values
val_ids   = pd.read_csv(proc / "val_ids.csv")["id"].astype(str).values
test_ids  = pd.read_csv(proc / "test_ids.csv")["id"].astype(str).values

TAB_FEATURES = pd.read_csv(proc / "tabular_scaler.csv")["feature"].tolist()

device = "cuda" if torch.cuda.is_available() else "cpu"
pin_memory = (device == "cuda")

(X_train_tab.shape, X_val_tab.shape, X_test_tab.shape, device)
```

```python
tab_model = XGBRegressor(
    n_estimators=4000,
    learning_rate=0.03,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1,
)

tab_model.fit(X_train_tab, y_train_log)

val_pred_log_tab = tab_model.predict(X_val_tab)

# Evaluate in price-space
val_true_price = np.expm1(y_val_log)
val_pred_price = np.expm1(val_pred_log_tab)

rmse_tab = float(np.sqrt(mean_squared_error(val_true_price, val_pred_price)))  # mean_squared_error is standard. [web:118]
r2_tab   = float(r2_score(val_true_price, val_pred_price))

print({"tabular_rmse": rmse_tab, "tabular_r2": r2_tab})
```

```python
imp_df = pd.DataFrame({"feature": TAB_FEATURES, "importance": tab_model.feature_importances_}) \
    .sort_values("importance", ascending=False)
imp_df.head(20)
```

```python
class MultiModalDS(Dataset):
    def __init__(self, ids, X_tab, y_log=None, images_dir="data/images/train", transform=None):
        self.ids = ids
        self.X_tab = X_tab
        self.y_log = y_log
        self.images_dir = Path(images_dir)
        self.transform = transform

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, i):
        pid = self.ids[i]
        img_path = self.images_dir / f"{pid}.png"
        img = Image.open(img_path).convert("RGB")
        if self.transform:
            img = self.transform(img)

        tab = torch.from_numpy(self.X_tab[i])  # float32

        if self.y_log is None:
            return img, tab, pid

        y = torch.tensor(float(self.y_log[i]), dtype=torch.float32)
        return img, tab, y, pid
```

```python
class MultiModalRegressor(nn.Module):
    def __init__(self, tab_dim: int, freeze_cnn: bool = True):
        super().__init__()

        weights = ResNet18_Weights.DEFAULT
        backbone = models.resnet18(weights=weights)  # resnet18 weights parameter is documented. [web:41]
        self.cnn = nn.Sequential(*list(backbone.children())[:-1])
        self.img_dim = 512

        if freeze_cnn:
            for p in self.cnn.parameters():
                p.requires_grad = False

        self.tab_mlp = nn.Sequential(
            nn.Linear(tab_dim, 256),
            nn.ReLU(),
            nn.BatchNorm1d(256),
            nn.Dropout(0.1),
            nn.Linear(256, 128),
            nn.ReLU(),
        )

        self.head = nn.Sequential(
            nn.Linear(self.img_dim + 128, 256),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(256, 1),
        )

    def forward(self, img, tab):
        img_feat = self.cnn(img).flatten(1)
        tab_feat = self.tab_mlp(tab)
        out = self.head(torch.cat([img_feat, tab_feat], dim=1)).squeeze(1)
        return out
```

```python
weights = ResNet18_Weights.DEFAULT
img_transform = weights.transforms()  # standard transforms attached to pretrained weights. [web:41]

ds_train = MultiModalDS(train_ids, X_train_tab, y_log=y_train_log, images_dir="data/images/train", transform=img_transform)
ds_val   = MultiModalDS(val_ids,   X_val_tab,   y_log=y_val_log,   images_dir="data/images/train", transform=img_transform)
ds_test  = MultiModalDS(test_ids,  X_test_tab,  y_log=None,        images_dir="data/images/test",  transform=img_transform)

dl_train = DataLoader(ds_train, batch_size=32, shuffle=True,  num_workers=2, pin_memory=pin_memory)
dl_val   = DataLoader(ds_val,   batch_size=32, shuffle=False, num_workers=2, pin_memory=pin_memory)
dl_test  = DataLoader(ds_test,  batch_size=32, shuffle=False, num_workers=2, pin_memory=pin_memory)
```

```python
from pathlib import Path
from PIL import Image

img_dir = Path("data/images/train")

missing = []
bad = []
for pid in train_ids[:2000]:  # sample first 2k
    p = img_dir / f"{pid}.png"
    if not p.exists():
        missing.append(pid)
        continue
    try:
        Image.open(p).convert("RGB")
    except Exception:
        bad.append(pid)

print("missing_images_sample:", len(missing))
print("corrupt_images_sample:", len(bad))
print("example_missing:", missing[:5])
print("example_corrupt:", bad[:5])
```

```python
out_dir = Path("outputs")
(out_dir / "checkpoints").mkdir(parents=True, exist_ok=True)
best_path = out_dir / "checkpoints" / "mm_best.pt"

model = MultiModalRegressor(tab_dim=X_train_tab.shape[1], freeze_cnn=True).to(device)

# HuberLoss is robust: combines L2 for small errors and delta-scaled L1 for large errors. [web:167]
loss_fn = nn.HuberLoss(delta=0.5)  # tune delta later if needed. [web:167]

def eval_mm():
    model.eval()
    y_true_log, y_pred_log = [], []
    with torch.no_grad():
        for img, tab, y, _ in dl_val:
            img = img.to(device, non_blocking=True)
            tab = tab.to(device, non_blocking=True)
            pred = model(img, tab).detach().cpu().numpy()
            y_pred_log.extend(pred.tolist())
            y_true_log.extend(y.numpy().tolist())

    y_true_price = np.expm1(np.array(y_true_log))
    y_pred_price = np.expm1(np.array(y_pred_log))

    rmse = float(np.sqrt(mean_squared_error(y_true_price, y_pred_price)))  # [web:118]
    r2 = float(r2_score(y_true_price, y_pred_price))
    return rmse, r2

best_rmse = float("inf")

# ---- Stage 1: warmup (CNN frozen) ----
opt = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=1e-3, weight_decay=1e-4)
WARMUP_EPOCHS = 3

for epoch in range(1, WARMUP_EPOCHS + 1):
    model.train()
    for img, tab, y, _ in dl_train:
        img = img.to(device, non_blocking=True)
        tab = tab.to(device, non_blocking=True)
        y   = y.to(device, non_blocking=True)

        pred = model(img, tab)
        loss = loss_fn(pred, y)

        opt.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step()

    rmse_mm, r2_mm = eval_mm()
    print({"stage": "warmup_frozen", "epoch": epoch, "tabular_rmse": rmse_tab, "multimodal_rmse": rmse_mm, "tabular_r2": r2_tab, "multimodal_r2": r2_mm})

    if rmse_mm < best_rmse:
        best_rmse = rmse_mm
        torch.save(model.state_dict(), best_path)

# ---- Stage 2: finetune (unfreeze CNN with tiny LR) ----
for p in model.cnn.parameters():
    p.requires_grad = True

opt = torch.optim.AdamW([
    {"params": model.cnn.parameters(), "lr": 1e-5},
    {"params": list(model.tab_mlp.parameters()) + list(model.head.parameters()), "lr": 3e-4},
], weight_decay=1e-4)

FINETUNE_EPOCHS = 5
for epoch in range(1, FINETUNE_EPOCHS + 1):
    model.train()
    for img, tab, y, _ in dl_train:
        img = img.to(device, non_blocking=True)
        tab = tab.to(device, non_blocking=True)
        y   = y.to(device, non_blocking=True)

        pred = model(img, tab)
        loss = loss_fn(pred, y)

        opt.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        opt.step()

    rmse_mm, r2_mm = eval_mm()
    print({"stage": "finetune_unfrozen", "epoch": epoch, "tabular_rmse": rmse_tab, "multimodal_rmse": rmse_mm, "tabular_r2": r2_tab, "multimodal_r2": r2_mm})

    if rmse_mm < best_rmse:
        best_rmse = rmse_mm
        torch.save(model.state_dict(), best_path)

print("best multimodal rmse:", best_rmse, "saved:", best_path)
```

```python
model.load_state_dict(torch.load(best_path, map_location=device))
model.eval()

preds_price = []
ids_out = []

with torch.no_grad():
    for img, tab, pid in dl_test:
        img = img.to(device, non_blocking=True)
        tab = tab.to(device, non_blocking=True)

        pred_log = model(img, tab).detach().cpu().numpy()
        pred_price = np.expm1(pred_log)

        preds_price.extend(pred_price.tolist())
        ids_out.extend(pid.tolist())

final = pd.DataFrame({"id": ids_out, "predicted_price": preds_price})
final.to_csv("outputs/predictions.csv", index=False)
final.head()
```
